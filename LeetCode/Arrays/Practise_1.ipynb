### Right Rotate an Array by One Place

```python
nums=[5,-2,3,8,9,11,15,12,2]

nums[:]=[nums[-1]]+nums[0:len(nums)-1]

print(nums)
```

### Right Rotate an Array by K Places

```python
nums=[5,-2,3,8,9,11,15,12,2]
length=len(nums)

times=int(input("Tell me how many times do you want the rotation to be done: "))
rotation=times%length

nums[:]=nums[length-rotation:]+nums[:length-rotation]

print(nums)

# Time Complexity -> o(n)
# Space Complexity ->o(1)
```

### Longest Consecutive Sequence

```python
# Better

nums=[5,-2,3,8,9,11,15,12,2,4]

nums.sort()

longest=0
count=0
last_smaller=float("-inf")

for i in range(len(nums)):
    num=nums[i]

    if num - 1 == last_smaller:
        count+=1
        last_smaller=num
    else:
        count=1
        last_smaller=num

    longest=max(longest,count)

print(longest)

# Time Complexity -> o(NlogN + N) == o(N)
# Space Complexity -> o(1)
```

```python
# Optimal

Unique_elements=set(nums)

longest = 0
count=0
for num in Unique_elements:
    if num - 1 not in Unique_elements:
        X=num
        count=1
        while X + 1 in Unique_elements:
            X+=1
            count+=1

        longest=max(longest,count)

print(longest)

# Time Complexity -> O(3N)
# Space Complexity -> O(N)
```

### 4 Sum

```python
# Better
def four_sum(num,target:int) :
    length = len(num)

    if length < 4:
        return []

    my_set=set()

    for i in range(length):
        for j in range(i+1,length):
            hash_set=set()
            for k in range(j+1,length):
                fourth=target-(num[i]+num[j]+num[k])

                if fourth in hash_set:
                    temp=[num[i],num[j],num[k],fourth]
                    temp.sort()
                    my_set.add(tuple(temp))

                hash_set.add(num[k])

    result=[list(ans) for ans in my_set]

    return result

# Time Complexity -> O(N^3)
```

```python
print(four_sum([1,5,8,7,3,4,2,8,-2,-1,-8],0))
```
