### Check if a String is Palindrome or Not

```python
def palindrome(left:int,right:int,string:str):
    if string[left]!=string[right]:
        return False

    if left>=right:
        return True

    palindrome(left+1,right-1,string)

    return True

print(palindrome(0,3,"abcd"))
```

### Find the Fibonacci Number

```python
def fibo(num):
    if num==1:
        return 1

    if num == 0:
        return 0

    return fibo(num-1)+fibo(num-2)

print(fibo(6))
```

### Selection Sort

```python
nums=[5,7,8,4,1,6,9,2]

def selection_sort(num:list[int])->list[int]:
    for i in range(len(num)):
        min_index=i
        for j in range(i+1,len(num)):
            if num[j]<num[min_index]:
                min_index=j
        num[i],num[min_index]=num[min_index],num[i]

    return num

print(selection_sort(nums))
```

```python
def bubble_sort(num:list[int])->list[int]:
    for i in range(len(num)-1):
        for j in range(len(num)-i-1):
            if num[j]>num[j+1]:
                num[j],num[j+1]=num[j+1],num[j]

    return num
nums = [5, 1, 4, 2, 8]
print(bubble_sort(nums))


# Time Complexity -> o(n^2)
# Space Complexity -> o(1)
```

```python
def insertion_sort(num:list[int])->list[int]:
    for i in range(len(num)):
        j=i-1
        key=num[i]

        while j>=0 and num[j] > key :
            num[j+1]=num[j]
            j-=1
        num[j+1]=key
    return num

print(insertion_sort(nums))


# Time Complexity -> o(n^2)
```

```python
def merge_array(left:list[int],right:list[int])->list[int]:
    l_pointer=0
    r_pointer=0
    result=[]
    while l_pointer<len(left) and r_pointer<len(right):
        if left[l_pointer]<=right[r_pointer]:
            result.append(left[l_pointer])
            l_pointer+=1
        else:
            result.append(right[r_pointer])
            r_pointer+=1

    while l_pointer<len(left):
        result.append(left[l_pointer])
        l_pointer+=1

    while r_pointer<len(right):
        result.append(right[r_pointer])
        r_pointer+=1

    return result
```

```python
def merge_sort(num:list[int])->list[int]:
    if len(num)<=1:
        return num

    mid = len(num)//2
    left=merge_sort(num[:mid])
    right=merge_sort(num[mid:])
    return merge_array(left,right)
nums = [5, 1, 4, 2, 8]
print(merge_sort(nums))

# Time Complexity -> o(NlogN)
# Space Complexity -> o(N)
```

```python
def partition(num:list[int],low,high):
    pivot=num[low]

    i=low
    j=high
    while i<j:
        while i<=high-1 and num[i]<=pivot:
            i+=1
        while j>=low + 1 and num[j]>=pivot:
            j-=1

        if i< j:
            num[i],num[j]=num[j],num[i]

    num[low],num[j]=num[j],num[low]

    return j
```

```python
def quick_sort(num:list[int],low:int,high:int)-> None:
    if low<high:
        p_index=partition(num,low,high)

        quick_sort(num,low,p_index-1)
        quick_sort(num,p_index+1,high)
```

```python
nums = [5, 1, 4, 2, 8]
quick_sort(nums,0,len(nums)-1)
print(nums)


# Time Complexity -> o(NlogN)
# Space Complexity -> o(1)
```
