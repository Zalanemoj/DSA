# Check if a Number is Palindrome or Not

```python
from math import *
n=1221
num=n
reversed_num=0

while num>0:
    last_digit=num%10
    reversed_num=reversed_num*10+last_digit
    num=num//10

if reversed_num==n:
    print('Number is Palindrome')
else:
    print('Number is not a Palindrome')
```

# Armstrong Number

```python
num=int(input('Enter the number'))
n=num
total=0

def count_len(num):
    return int(log10(num)+1)

length=count_len(n)
while n>0:
    digit=n%10
    total+=digit**length
    n=n//10

if total==num:
    print('Armstrong Number')
else:
    print('Not an Armstrong Number')
```

# Print All Factors of a Given Number

### Brut force approch

```python
num=int(input('Enter the number'))

factors=[]

for i in range(1,num+1):
    if num%i==0:
        factors.append(i)

print(f'The factors of the number are {factors}')
```

### More practical(Time optimised)

```python
num=int(input('Enter the number'))

factors=[]

for i in range(1,int((num+1)/2)):
    if num%i==0:
        factors.append(i)
factors.append(num)
print(f'The factors of the number are {factors}')
```

### Most optimized

```python
num=int(input('Enter the number'))

factors=[]

for i in range(1,int(sqrt(num)+1)):
    if num%i==0:
        factors.append(i)
        if i!=num//i:
            factors.append(int(num//i))

factors.sort()
print(f'The factors of the number are {factors}')
```
