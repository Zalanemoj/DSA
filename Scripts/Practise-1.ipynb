## Find the Largest Element in an Array

### Method 1 ----> loop method
Time Complexity o(n)\
Space Complexity o(1)

```python
nums=[10,15,18,26,-99,-85,36,57]

largest_num=nums[0]

for i in nums:
    if i>largest_num:
        largest_num=i

print('The largest number is ',largest_num,sep='\n')
```

### Method 2 ----> Sorting

```python
nums=[10,15,18,26,-99,-85,36,57]

nums.sort()

largest_num=nums[len(nums)-1]
print('The largest number is ',largest_num,sep='\n')
```

## Find the Second Largest Element in an Array Without Sorting

```python
nums=[10,15,18,26,-99,-85,36,57]

largest_num=nums[0]

for i in nums:
    if i>largest_num:
        largest_num=i

second_largest=nums[0]

for i in nums:
    if second_largest < i < largest_num:
        second_largest=i

print('The second largest number is ',second_largest,sep='\n')
```

```python
nums.sort()

print(nums)
```

### Optimal Solution
Time complexity o(n)

```python
nums=[10,15,18,26,-99,-85,36,57]

largest_num=nums[0]
second_largest=nums[0]

for i in nums:
    if i>largest_num:
        if largest_num>second_largest:
            second_largest=largest_num
        largest_num=i

print('The second largest number is ',second_largest,sep='\n')
```

## Check if an Array is Sorted
Time Complexity o(n)\
Space Complexity o(1)

```python
nums=[10,15,18,26,-99,-85,36,57]
sorted_list=True
for i in range(0,len(nums)-2):
    if nums[i]>nums[i+1]:
        sorted_list=False
        break
print('The given array is sorted or not:',sorted_list)
```

## Remove Duplicates from a Sorted Array

```python
nums=[10,15,18,26,-99,-85,36,57,57,26,13,1,30,30,10,15]

nums.sort()

print('The sorted array is',nums,sep='\n')

array_without_duplicate=set(nums)

print('Array after removal of duplicate ',array_without_duplicate,sep='\n')
```

```python
nums = [10,15,18,26,-99,-85,36,57,57,26,13,1,30,30,10,15]
nums.sort()

print('The sorted array is', nums, sep='\n')

i = 0
while i < len(nums) - 1:
    if nums[i] == nums[i + 1]:
        nums.pop(i)
    else:
        i += 1

print('The array without duplicate is', nums, sep='\n')
```

## Right Rotate an Array by One Place

```python
nums=[10,15,18,26,-99,-85,36,57]

length:int=len(nums)
nums[:]=[nums[length-1]]+nums[0:length-1]

print(nums)
```

```python
for i in range(length-2,-1,-1):
    if nums[i]==nums[i-1]:
        nums.pop(i)

print(nums)
```

## Right Rotate an Array by K Places
Time Complexity o(n)\
Space complexity o(1)

```python
nums=[10,15,18,26,-99,-85,36,57]
rotation=int(input('Enter the number of times you want to rotate the array: '))
length:int=len(nums)
rotation = rotation % length
nums[:]=nums[length-rotation:]+nums[:length-rotation]

print(nums)
```

### Optimal Solution

```python
nums=[10,15,18,26,-99,-85,36,57]
rotation=int(input('Enter the number of times you want to rotate the array: '))
length:int=len(nums)
rotation = rotation % length

def reverse_array(num,left,right):
    if left>right:
        return
    num[left],num[right]=num[right],num[left]
    left+=1
    right-=1
    reverse_array(num,left,right)

left=0
right=length-1
reverse_array(nums,length-rotation,right)
reverse_array(nums,left,length-rotation-1)
reverse_array(nums,left,right)

print(nums)
```

## Move Zeros to the End of the List
Time Complexity o(2n)\
Space Complixity o(1)

```python
nums=[10,15,18,26,0,0,-99,-85,0,36,57,0]

print('The list Before insertion of zeros at last:',nums,sep='\n')

count=0
i=0

while i<len(nums):
    if nums[i]==0:
        count+=1
        nums.pop(i)
    else:
        i+=1

for i in range(1,count+1):
    nums.append(0)

print('The list after insertion of zeros at last:',nums,sep='\n')
```
