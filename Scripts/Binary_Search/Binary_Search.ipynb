## Iterative method

Time complexity --->o(logn)\
Space complexity ---->o(1)

```python
from numpy.ma.core import minimum

nums=[2,4,6,7,9,11,18,19]

target=11

def Binary_Serch(num,target):
    low=0
    high=len(nums)-1
    while low<=high:
        mid=(low+high)//2
        if num[mid]==target:
            return print('The index where the number exist is:',mid,sep='\n')
        elif num[mid]<target:
            low=mid+1
        else:
            high=mid-1
    return None

Binary_Serch(nums,target)
```

## Recursive Solution

```python
def binary_recursive(num,low,high,target):
    if low>high:
        return None
    mid=(low+high)//2
    if num[mid]==target:
        return print('The index where the number exist is:',mid,sep='\n')
    elif num[mid]<target:
        return binary_recursive(num,mid+1,high,target)
    else:
        return binary_recursive(num,low,mid-1,target)

binary_recursive(nums,0,len(nums)-1,target)
```
