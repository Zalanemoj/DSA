## 3Sum Problem

```python
nums = [-1, 0, 1, 2, -1, -4]
```

### Brute Force Solution (Triple Nested Loops)

Time Complexity ---> o(n^3)\
Space Complexity ---> o(No. of elements)

```python
class Solution:
    @staticmethod
    def three_sum(num):
        result=[]

        for i in range(0, len(num)-2):
            for j in range(i+1, len(num)-1):
                for k in range(j+1, len(num)):
                    if num[i]+num[j]+num[k]==0:
                        temp=[num[i],num[j],num[k]]
                        temp.sort()
                        if temp not in result:
                            result.append(temp)
        return result

print(Solution.three_sum(nums))
```

### Better Solution (Hash-Set per Fixed Index)

Time Complexity ---> o(n^2)\
Space Complexity ---> o(No. of Triplets)

```python
class Solution:
    @staticmethod
    def three_sum(num):
        result=[]
        for i in range(len(num)-1):
            my_set=set()
            for j in range(i+1, len(num)):
                third=-(num[i]+num[j])
                if third in my_set:
                    temp=[num[i],num[j],third]
                    temp.sort()
                    if temp not in result:
                        result.append(temp)
                my_set.add(num[j])
        return result
```

```python
print(Solution.three_sum(nums))
```

### Optimal Solution (Sorting + Two Pointers)

Time Complexity ---> o(n^2)\
Space Complexity ---> o(No. of elements)

```python
class Solution:
    @staticmethod
    def three_sum_optimal(num):
        result=[]

        num.sort()
        for i in range(len(num)):
            if i!=0 and num[i] == num[i-1]:
                continue

            j=i+1
            k=len(num)-1
            while j<k:
                if num[i]+num[j]+num[k]>0:
                    k-=1
                elif num[i]+num[j]+num[k]<0:
                    j+=1
                else:
                    temp=[num[i],num[j],num[k]]
                    result.append(temp)

                    k-=1
                    j+=1
                    while j<k and num[j]==num[j+1]:
                        j+=1
                    while j<k and num[k]==num[k-1]:
                        k-=1
        return result

print(Solution.three_sum_optimal(nums))
```

## 4Sum Problem

### Optimal Solution (Sort + Two-Pointer Technique)

```python
nums = [1, 0, -1, 0, -2, 2]

class Solution:
    @staticmethod
    def four_sum(num):
        result=[]
        num.sort()

        length=len(num)

        for i in range(length):
            if i!=0 and num[i]==num[i-1]:
                continue

            for j in range(i+1,length):
                if j>(i+1) and num[j]==num[j-1]:
                    continue

                k=j+1
                l=length-1

                while k<l:
                    total=num[i]+num[j]+num[k]+num[l]
                    if total>0:
                        l-=1
                    elif total < 0:
                        k+=1
                    else:
                        temp=[num[i],num[j],num[k],num[l]]

                        if temp not in result:
                            result.append(temp)

                        k+=1
                        l-=1

                        while k<l and num[k]==num[k-1]:
                            k+=1

                        while k<l and num[l]==num[l+1]:
                            l-=1
        return result

print(Solution.four_sum(nums))
```
