# Sorting

## Selection Sort
In this we keep a trace on minimum index and an index and changes the index position with minimum index by iterating an I and J loop.

### Using loop method

```python
nums=[10,25,2,3,7,14,18,19,15,1,5]

print('The list before selection sort is:',nums,sep='\n')

for i in range(0,len(nums)-1):
    min_index=i
    for j in range(i+1,len(nums)):
        if nums[j] < nums[min_index]:
            min_index=j
    nums[i],nums[min_index]=nums[min_index],nums[i]

print('The list after selection sort is:',nums,sep='\n')
```

## Bubble Sort(Adjacent Sort)
It works on the phenomena that if the second element from now is greater than Then you have to swap both of them and by doing that you would get assorted array at the end.


```python
nums=[10,25,2,3,7,14,18,19,15,1,5]

print('The list before Bubble sort is:',nums,sep='\n')

for i in range(len(nums)-2,-1,-1):
    for j in range(0,i+1):
        if nums[j]>nums[j+1]:
            nums[j],nums[j+1]=nums[j+1],nums[j]

print('The list after Bubble sort is:',nums,sep='\n')
```

### Time Complexity o(n)

```python
nums=[10,25,2,3,7,14,18,19,15,1,5]

print('The list before Bubble sort is:',nums,sep='\n')

for i in range(len(nums)-2,-1,-1):
    is_swap=False
    for j in range(0,i+1):
        if nums[j]>nums[j+1]:
            nums[j],nums[j+1]=nums[j+1],nums[j]
            is_swap=True
    if is_swap==False:
        break

print('The list after Bubble sort is:',nums,sep='\n')
```

## Insertion sort
It is a method of sorting in which we replace his by copping an and replacing or shifting the position buy one and adding the element where it is needed to be added by iterating a loop.

```python
nums=[10,25,2,3,7,14,18,19,15,1,5]

print('The list before Insertion sort is:',nums,sep='\n')

for i in range(0,len(nums)):
    value=nums[i]
    for j in range(i-1,-1,-1):
        if value<nums[j]:
            nums[j+1]=nums[j]
        elif value>nums[j]:
            nums[j+1]=value
            break
    else:
        nums[0]=value

print('The list after Insertion sort is:',nums,sep='\n')
```

## Merge sort(Divide and merge)
In this algorithm we first of all divide the given list into parts and then merge it bases on the sorting.

Time complexity ----> o(nlogn)\
Space complexity ----> o(n)

```python
import warnings
warnings.filterwarnings('ignore')
def merge_array(n,m):
    n_index=0
    m_index=0
    result=[]
    while n_index<len(n) and m_index<len(m):
        if n[n_index]<=m[m_index]:
            result.append(n[n_index])
            n_index+=1
        else:
            result.append(m[m_index])
            m_index+=1
    while n_index<len(n):
        result.append(n[n_index])
        n_index+=1

    while m_index<len(m):
        result.append(m[m_index])
        m_index+=1

    return result

def divide_array(num):
    if len(num)<=1:
        return num
    mid=len(num)//2
    left_half=num[0:mid]
    right_half=num[mid:]
    left_half=divide_array(left_half)
    right_half=divide_array(right_half)
    return merge_array(left_half,right_half)

nums=[1,5,8,9,7,2,3,6,4,14,18,24,10,7,36,25,14,18]
print('The list before Merge sort is:',nums,sep='\n')

list_after_sorting=divide_array(nums)

print('The list after Merge sort is:',list_after_sorting,sep='\n')
```

## Quick Sort
In this method of sorting we take a pivot and Assign the pivot its correct position and to the left of it we do like all the smaller numbers and to the right of it all the larger numbers and we do it iteratively in order to get the correct sort.

```python
def partition(num,low,high):
    pivot=num[low]
    i=low
    j=high
    while i<j:
        while i<=high-1 and num[i]<=pivot:
            i+=1
        while j>=low +1 and num[j]>pivot:
            j-=1
        if i<j:
            num[i],num[j]=num[j],num[i]
    num[low],num[j]=num[j],num[low]
    return j

def quick_sort(num,low,high):
    if low<high:
        p_index=partition(num,low,high)
        quick_sort(num,low,p_index-1)
        quick_sort(num,p_index+1,high)

arr = [64, 34, 25, 12, 22, 12, 12, 11, 90]
print('The list before Quick sort is:',arr,sep='\n')
quick_sort(arr,0,len(arr)-1)
print('The list after Quick sort is:',arr,sep='\n')
```
