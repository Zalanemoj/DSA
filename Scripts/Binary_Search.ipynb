## Iterative method

Time complexity --->o(logn)\
Space complexity ---->o(1)

```python
nums=[2,4,6,7,9,11,18,19]

target=11

def Binary_Serch(num,target):
    low=0
    high=len(nums)-1
    while low<=high:
        mid=(low+high)//2
        if num[mid]==target:
            return print('The index where the number exist is:',mid,sep='\n')
        elif num[mid]<target:
            low=mid+1
        else:
            high=mid-1
    return None

Binary_Serch(nums,target)
```

## Recursive Solution

```python
def binary_recursive(num,low,high,target):
    if low>high:
        return None
    mid=(low+high)//2
    if num[mid]==target:
        return print('The index where the number exist is:',mid,sep='\n')
    elif num[mid]<target:
        return binary_recursive(num,mid+1,high,target)
    else:
        return binary_recursive(num,low,mid-1,target)

binary_recursive(nums,0,len(nums)-1,target)
```

Lower bound ----> Smallest index such nums of I is greater than equal to target.\
nums[i]>=target

Upper bound ----> Smallest index such that nums of I is greater than target.\
nums[i]>target

```python
nums=[1,1,1,2,3,3,5,6,7,7,7,9,12,12,13]

def lower_bound(num,target):
    low=0
    high=len(num)-1
    lowerBound=-1
    while low<=high:
        mid=(low+high)//2
        if num[mid]>=target:
            lowerBound=mid
            high=mid-1
        else:
            low=mid+1
    return print('The index where the number exist is:',lowerBound,sep='\n')

lower_bound(nums,12)
```

```python
def upper_bound(num,target):
    high=len(num)-1
    print('The length of the given list is:',len(num),sep='\n')
    upperBound=len(nums)+1
    low=0
    while low<=high:
        mid=(low+high)//2
        if num[mid]<target:
            upperBound=mid+1
            low=mid+1
        else:
            high=mid-1
    return print('The index where the number exist is:',upperBound,sep='\n')

upper_bound(nums,13)
```

## Search Insert Position

```python
# If element is present in the list given written the index of it otherwise tell the index where it can be inserted.
nums=[1,3,4,5,8,9,14,15,19,20,21]

print('The length of the given list is:',len(nums),sep='\n')

def Search_Position(num,target):
    low=0
    high=len(nums)-1
    position=-1
    while low<=high:
        mid=(low+high)//2
        if num[mid]==target:
            position=mid
            return print('The index where the number exist is:',position,sep='\n')
        elif num[mid]>target:
            high=mid-1
            position=mid
        else:
            low=mid+1
            position=mid
    return print('The index where the number should be placed is:',position,sep='\n')
Search_Position(nums,25)
```
