## Print 1 to n using recursion

```python
n=int(input('Enter the no.:'))

def number_printer(i,num):
    if i>num:
        return
    print(i)
    return number_printer(i+1,num)

number_printer(1,n)
```

### Tail tracking

```python
n=int(input('Enter the no.:'))

def number_printer(i,num):
    if i>num:
        return

    number_printer(i+1,num)
    return print(i)

number_printer(1,n)
```

## Print n to 1 using head tracking

```python
n=int(input('Enter the no.:'))

def number_printer(i,num):
    if i>num:
        return
    print(num)
    return number_printer(i,num-1)

number_printer(1,n)
```

### Tail tracking

```python
n=int(input('Enter the no.:'))

def number_printer(i,num):
    if i>num:
        return

    number_printer(i,num-1)
    return print(num)

number_printer(1,n)
```

## Factorial of a number

```python
n=int(input('Enter the no.:'))
# import warnings
# warnings.filterwarnings('ignore')
import errno
def factorial(num):
    if num<0:
        raise ValueError(' The factorial of a number is not possible because it is an invalid number or it is a very large number that you have given.')
    if num==0 or num==1:
        return 1
    return num*factorial(num-1)

value=factorial(n)
print(f'The factorial of the number {n} is {value}')
```

## Reverse an array using recursion

```python
n = list((5,2,3,5,7,9,11))

print('The original list is:', n, sep='\n')

def reverse(pos, arr):
    if pos >= len(arr) // 2:
        return
    arr[pos], arr[len(arr)-1-pos] = arr[len(arr)-1-pos], arr[pos]
    reverse(pos + 1, arr)

reverse(0, n)

print('The reversed list is:', n, sep='\n')
```

### Reverse from n left and n right

```python
right=int(input('Enter the position from the right:'))
left=int(input('Enter the position from the left:'))

n = list((5,2,3,5,7,9,11,15,17,18,19,15,14,17,12,10,16,14,13))

print('The original list is:', n, sep='\n')

def reverse_pos(arr,l,r):
    if l >= r:
        return
    arr[l],arr[r]=arr[r],arr[l]
    reverse_pos(arr,l+1,r-1)

reverse_pos(n,left,right)

print('The reversed list is:', n, sep='\n')
```

## Check if a String is Palindrome or Not

```python
n = input("Enter the string: ")

n_list = list(n)

def reverse_pos(arr, l, r):
    if l >= r:
        return
    arr[l], arr[r] = arr[r], arr[l]
    reverse_pos(arr, l+1, r-1)

reverse_pos(n_list, 0, len(n_list)-1)
reversed_string = ''.join(n_list)

if n == reversed_string:
    print('The given string is a Palindrome')
else:
    print('Not a palindrome')
```

## Find the Fibonacci Number

```python
n=int(input('Enter the no.:'))

def fibonacci(num):
    if num==0:
        return 0

    if num==1:
        return 1

    if num<0 or num>10**5:
        raise ValueError('The provided number is either negative or it is much more greater.')
    return fibonacci(num-1)+fibonacci(num-2)

value=fibonacci(n)

print('The value of Fibonacci at the particular input is:',value)
```
